### Creación de Vectores para su posterior uso.
Crearemos los vectores que querramos usar según nuestras necesidades, para después poder visualizarlos al imprimirlos.

```python
# Ejercicio 2: Creación de vectores
v1 = np.array([3, 4, 5])
v2 = np.array([1, 2, 3])
v3 = np.array([0, -2, 1])

print("v1 =", v1)
print("v2 =", v2)
print("v3 =", v3)
print("\n" + "="*30)

suma = v1 + v2
print("v1 + v2 =", suma)

operacion = 2*v1 - v3
print("2*v1 - v3 =", operacion)
```

### Creación de pequeña función para el calculo de la norma de un vector y su comparación con la proporcionada por las librerias.

```python
# Función para calcular norma
def norma_vector(vector):
    cuadrados = [e**2 for e in vector]
    return sum(cuadrados)**0.5

print("Norma de v1:")
print(f"  Mi función: {norma_vector(v1):.4f}")
print(f"  np.linalg.norm: {np.linalg.norm(v1):.4f}")
```

Código para calcular el producto punto y el ángulo de dos vectores, además de un pequeño ejemplo de su funcionamiento.

```python
# Producto punto y ángulo
a = np.array([2, 5])
b = np.array([3, 1])

producto_punto = np.dot(a, b)
print(f"a · b = {producto_punto}")

cos_theta = producto_punto / (np.linalg.norm(a) * np.linalg.norm(b))
angulo_rad = np.arccos(cos_theta)
angulo_grados = angulo_rad * 180 / np.pi

print(f"Ángulo entre a y b: {angulo_rad:.4f} radianes = {angulo_grados:.2f}°")
```

Código para graficar vectores 2D y breve ejemplo de su uso además de visualización para la suma de vectores.

```python
# Función para graficar vectores 2D
def plot_vector2d(vector2d, origin=[0, 0], **options):
    return plt.arrow(origin[0], origin[1], vector2d[0], vector2d[1],
              head_width=0.2, head_length=0.3, length_includes_head=True,
              **options)

print("Función de graficado definida")

# Graficar vectores
a = np.array([2, 5])
b = np.array([3, 1])
suma = a + b

plt.figure(figsize=(8, 6))

plot_vector2d(a, color='red', linewidth=2)
plot_vector2d(b, color='blue', linewidth=2)
plot_vector2d(suma, color='green', linewidth=2)

# Líneas punteadas para la regla del paralelogramo
plot_vector2d(b, origin=a, color='blue', linestyle='dotted', alpha=0.5)
plot_vector2d(a, origin=b, color='red', linestyle='dotted', alpha=0.5)

plt.xlim(0, 9)
plt.ylim(0, 7)
plt.grid(True, alpha=0.3)
plt.title('Suma de vectores', fontsize=14)
plt.xlabel('x')
plt.ylabel('y')

plt.text(0.5, 3, 'a', color='red', fontsize=14)
plt.text(4, 0.5, 'b', color='blue', fontsize=14)
plt.text(2.5, 3, 'a+b', color='green', fontsize=14)

plt.show()
```
